```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def __str__(self):
        return f"Employee: {self.name}"

    def __repr__(self):
        return f"Employee('{self.name}', {self.salary})"


emp = Employee("John", 50000)

print(emp)        # uses __str__
print(repr(emp))  # uses __repr__
```

```python
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance
    def deposite(self,amount):
        self.amount = amount
        print("deposited {amount}.New {self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("withdraw {amount}. New balance : {self.balance}")
        else:
            print("Insufficient funds")

    def __str__(self):
        return f"Account[owner={self.owner}, balance={self.balance}]"
    
acc = BankAccount("Alice", 1000)

acc.deposite(500)
acc.withdraw(200)
acc.withdraw(2000)

print(acc)  
    
```

```python
def safe_read(filepath):
    try:
        with open(filepath, "r") as f:
            return f.read()
    except FileNotFoundError:
        return f"❌ Error: '{filepath}' not found."
    except PermissionError:
        return f"❌ Error: No permission to read '{filepath}'."
    except Exception as e:
        return f"❌ Unexpected error: {e}"

print(safe_read("Sample_fh.txt"))
print()
```
